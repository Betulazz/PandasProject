# 2. Matplotlib入门案例_绘制标准折线图

```python
# 需求：绘制北京某天1个小时内的气温折线图
import random
```

## 2.1 准备数据并绘制初始折线图

```python
# 1. 准备x轴（60分钟）和y轴（每分钟的温度）的数据
x = range(60)  # [0,60)
y_beijing = [random.uniform(15, 18) for _ in x]

# 2. 创建画布
plt.figure(figsize=(10, 5), dpi=80)

# 3. 开始绘图
plt.plot(x, y_beijing, color='red')

# 4. 具体绘图动作
plt.show()
```

## 2.2 添加x轴，y轴刻度信息

```python
# plt.xticks(x, **kwargs)
# 1. 准备x轴（60分钟）和y轴（每分钟的温度）的数据
x = range(60)  # [0,60)
y_beijing = [random.uniform(15, 18) for _ in x]

# 2. 创建画布
plt.figure(figsize=(10, 5), dpi=80)

# 3. 开始绘图
plt.plot(x, y_beijing, color='red')

# 4. 添加x轴和y轴的刻度信息
x_ticks_label = [f'11点{i}分' for i in x]
y_ticks = range(40)

plt.xticks(x[::5], x_ticks_label[::5])  # 参1：x轴刻度值 参2：x轴刻度值对应的标签
plt.yticks(y_ticks[::5])

# 5. 添加网格
plt.grid(linestyle=':', alpha=0.5)

# 6. 添加标题
plt.title('北京某天1个小时内的气温折线图', fontsize=20)
# 7. 添加x轴 y轴标签
plt.xlabel('时间', fontsize=12)
plt.ylabel('温度', fontsize=12)
# 8. 保存图片
plt.savefig('./my_picture.png')
# 9. 具体绘图动作
plt.show()
```

# 3. 在一个画布中多次plot()绘制图形

```python
# 需求：在一个画布中，绘制两个城市的温度图
# 1. 准备数据
x = range(60)
y_beijing = [random.uniform(5, 18) for _ in x]
y_shanghai = [random.uniform(10, 15) for _ in x]

# 2. 创建画布
plt.figure(figsize=(20, 8), dpi=80)

# 3. 具体的绘制动作
# 3.1 绘制出两个折线
plt.plot(x, y_beijing, label='北京')
plt.plot(x, y_shanghai, label='上海')

# 3.2 创建x的刻度值，及x轴 y轴的刻度范围
x_ticks_label = [f'11点{i}分' for i in x]
y_ticks = range(20)
plt.xticks(x[::5], x_ticks_label[::5])
plt.yticks(y_ticks[::2])

# 3.3 标题 x轴 y轴
plt.title('北京上海1个小时内的气温折线图', fontsize=20)
plt.xlabel('时间', fontsize=12)
plt.ylabel('温度', fontsize=12)

# 3.4 添加网格
plt.grid(linestyle='--', alpha=0.5)

# 3.5 添加图例
# plt.legend(loc='best')
plt.legend(loc=0)
# 4. 绘制图形
plt.show()
```

# 4. 一个画布，多个坐标轴 -> 面向对象绘图方式

```python
# 需求：在一个画布中，绘制两个城市的温度图
# 1. 准备数据
x = range(60)
y_beijing = [random.uniform(5, 18) for _ in x]
y_shanghai = [random.uniform(10, 15) for _ in x]

# 2. 创建画布
# 以元组的方式返回结果，即：(画布对象，坐标轴对象)
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 8))

# 3. 具体的绘制动作
# 3.1 绘制出两个折线
axes[0].plot(x, y_beijing, label='北京')
axes[1].plot(x, y_shanghai, label='上海')

# 3.2 创建x的刻度值，及x轴 y轴的刻度范围
x_ticks_label = [f'11点{i}分' for i in x]
y_ticks = range(20)
axes[0].set_xticks(x[::5], x_ticks_label[::5])
axes[0].set_yticks(y_ticks[::2])

axes[1].set_xticks(x[::5])
axes[1].set_xticklabels(x_ticks_label[::5])
axes[1].set_yticks(y_ticks[::2])

# 3.3 标题 x轴 y轴
axes[0].set_title('北京上海1个小时内的气温折线图', fontsize=20)
axes[0].set_xlabel('时间', fontsize=12)
axes[0].set_ylabel('温度', fontsize=12)

axes[1].set_title('北京上海1个小时内的气温折线图', fontsize=20)
axes[1].set_xlabel('时间', fontsize=12)
axes[1].set_ylabel('温度', fontsize=12)

# 3.4 添加网格
axes[0].grid(linestyle='--', alpha=0.5)
axes[1].grid(linestyle='-.', alpha=0.5)

# 3.5 添加图例
# plt.legend(loc='best')
axes[0].legend(loc='best')
axes[1].legend(loc='best')

# 保存图形
plt.savefig('./my_picture2.png')

# 4. 绘制图形
plt.show()
```

# 6. Matplotlib绘制其他常用图形

## 6.1 柱状图bar() 适用于对比，统计

```python
x = ['A', 'B', 'C', 'D']
y = [10, 6, 15, 21]

plt.figure(figsize=(10, 5))

plt.bar(x, y)

plt.grid(linestyle='--', alpha=0.5)
plt.title('柱状图', fontsize=20)
plt.xlabel('分类名', fontsize=12)
plt.ylabel('数量', fontsize=12)

plt.show()
```

## 6.2 直方图hist() 适用于连续的数据的 分布，统计

```python
data = np.random.randn(500)  # 生成符合正态分布的随机500个数

plt.figure(figsize=(10, 5), dpi=80)

plt.hist(data, bins=5)

plt.grid(linestyle='--', alpha=0.5)
plt.title('直方图', fontsize=20)

plt.show()
```

## 6.3 饼图pie() -> 适用于比例的统计

```python
sizes = [25, 35, 25, 15]
labels = ['Category A', 'Category B', 'Category C', 'Category D']
plt.pie(sizes, labels=labels, autopct='%1.1f%%')  # 显示百分比
plt.title('饼图')
plt.show()
```

## 6.4 散点图 scatter() -> 适用于散点数据的统计 或者统计两列值之间的关系

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
plt.scatter(x, y, color='r')
plt.title('散点图')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
