```python
# 1. 完成需求
# 需求1：筛选出列值 > 23的数据
df[df.open > 23]

# 需求2：筛选出open列值 > 23 且 < 24 的数据
df[(df.open > 23) & (df.open < 24)]  # 多组判断是小括号(元组)

df[(df['open'] > 23) & (df['open'] < 24)]  # 标准写法
```

- apply()函数 -> 可以执行自定义函数

```python
# 格式：df.apply(自定义函数，axis=0)
# 需求：同时获取到多列的最大值和最小值的差值
# 思路1：分解版
# 1. 自定义函数 my_func，接受 某列的数据，计算该列的最大值和最小值，返回差值
def my_func(col):
    return col.max() - col.min()


# 获取到open列和close列
df[['open', 'close']]
df.loc[:, ['open', 'close']]
df.iloc[:, [0, 2]]

# 3. 通过apply()函数，调用上述的自定义函数，作用到指定的列

df[['open', 'close']].apply(my_func, axis=1)  # axis=1 对行进行操作
df[['open', 'close']].apply(my_func, axis=0)  # axis=0 对列进行操作 默认为0
```

```python
# 思路2：合并版，通过lambda函数实现
df[['open', 'close']].apply(lambda col: col.max() - col.min())
```
