# 1. Pandas的增删改查操作 -> 增

```python
# 1. 准备数据
df = pd.read_csv('./data/1960-2019全球GDP数据.csv', encoding='gbk')
df
```

```python
# 2. 从原始数据集中拷贝一部分出来操作
df2 = df[:5].copy()
df2
```

```python
# 3. 新增列
# 思路1：直接赋值，新增一列
df2['c1'] = 23  # 固定值
df2['c2'] = ['张三', '李四', '王五', '赵六', '孙七']  # 传入列表
df2['c3'] = df2.year * 2  # 通过已有的列（Series）来计算新列的值


def my_fun1():
    return 25000


df2['c4'] = my_fun1()  #写法四：通过函数来计算新列的值
df2
```

```python
# 思路2：通过assign()的方式新增1列
df3 = df[:5].copy()
df3
```

```python
# 通过assign()函数新增1列
df3.assign(c1=23)
```

```python
# 通过assign()函数新增n列
df3.assign(
    c1=23,
    c2=['张三', '李四', '王五', '赵六', '孙七'],
    c3=df3.year * 2,
    c4=my_fun1()
)
```

```python
# 8. rank()函数，类似于sql中的窗口函数
df7 = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六'],
    '成绩': [100, 90, 90, 80]
})
df7
```

```python
# 9. rank()直接排名
df7.rank()  # 针对于每列都会排名
# 给df新增1列，用于显示排名
df7['成绩排名_min']=df7.成绩.rank(method='min',ascending=False)
df7['成绩排名_max']=df7.成绩.rank(method='max',ascending=False)
df7['成绩排名_dense']=df7.成绩.rank(method='dense',ascending=False)
df7['成绩排名_avg']=df7.成绩.rank(method='average',ascending=False)
df7['成绩排名_默认']=df7.成绩.rank(ascending=False)
df7
```
