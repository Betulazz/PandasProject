# 1. 扩展_Pandas绘图

```python
# 介绍：Pandas绘图可以基于df对象直接绘图，它的底层依赖Matplotlib
```

```python
# 1. 加载数据集，并设置第一列为索引列
df = pd.read_csv('./data/winemag-data_first150k.csv', index_col=0)
df
```

```python
# 2. 查看数据集
df.info()
df.describe()
df.columns
```

- 需求1：绘制图形，展示产葡萄酒最多的10个产地的信息

```python
# 3. 分组统计
# 思路1：根据产地（province）分组，基于points列（不含缺失值的列）进行count统计即可
df.groupby('province', as_index=False).points.count().sort_values('points', ascending=False).head(10)
# 思路2：value_count()值的个数，且会自动降序排列，相当于 groupby()+count()+sort_values
df['province'].value_counts().head(10)
```

```python
# 4. 针对于上述的结果，绘图展示
# df['province'].value_counts().head(10).plot()   # 折线图 语法糖
# df['province'].value_counts().head(10).plot.line()   # 同上
df['province'].value_counts().head(10).plot.bar(figsize=(10, 5), fontsize=15)  # 柱状图
```

```python
# 5. 最终写法，记录参数，字典形式
# step1：定义变量，记录参数，字典形式
text_kwargs = dict(figsize=(10, 5), color=['r', 'g', 'b', 'y', 'c', 'm', 'k', 'pink', 'grey'], fontsize=15)
# print(text_kwargs)

# step2：绘图，传参
df['province'].value_counts().head(10).plot.bar(**text_kwargs)
```

- 需求2：绘制图形，展示葡萄酒最多的10个产地的占比

```python
# 1. 计算 葡萄酒最多的10个产地
df['province'].value_counts().head(10)

# 2. 计算每个产地的占比，即：产地产的葡萄酒数量 / 总共的葡萄酒数量
# len(df)
df['province'].value_counts().head(10) / len(df)

# 3. 绘图
(df['province'].value_counts().head(10) / len(df)).plot.bar(**text_kwargs)
```

- 需求3：展示每个频分的葡萄酒种类（个数），即：80分 -> 多少种酒......

```python
df.points.value_counts().sort_index().plot.line(**text_kwargs)  # 折线图
df.points.value_counts().sort_index().plot.area(**text_kwargs)  # 面积图，就是把区域内进行填充
df.points.value_counts().sort_index().plot.bar(**text_kwargs)
```

# 2. 扩展_Seaborn绘图

```python
# Seaborn是一个数据可视化库，基于Matplotlib，提供了更高级的绘图功能
# Pandas中绘图直接写单词，如：hist()，Seaborn中绘图是图形名+plot()，例如：histplot()
```

```python
# 1. 加载数据集合
# df = pd.read_csv('./data/tips.csv')
df = sns.load_dataset('tips')  # 在线获取Seaborn自带的数据集
df
```

- 需求1：绘制图形，展示男女总账单的清空

```python
# 1. 创建画布
fig, ax = plt.subplots(figsize=(10, 5))

# 2. 绘制图形
# 参1：data：数据集 参2：x：x轴的列名 参3：hue：分组依据的列名
sns.histplot(data=df, x='total_bill', hue='sex')

# 3. 添加标题
ax.set_title('男女总账单的情况', fontsize=15)

# 4. 展示图形
plt.show()
```

- 散点图，scatter() regplot() jointplot()

```python
# 需求：描述总账单和小费的分布情况
# 思路1：散点图，scatter()
fig, ax = plt.subplots(figsize=(10, 5))

# sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex')

# 思路2：regplot()，可以绘制散点图，且可以绘制拟合回归线
# sns.regplot(data=df, x='total_bill', y='tip')

# 思路3：jointplot()
sns.jointplot(data=df, x='total_bill', y='tip', kind='hex')  # kind=hex 蜂巢图

ax.set_title('总账单和小费的分布情况')

plt.show()
```

- 箱线图，boxplot()，violinplot()

```python
# 绘制箱线图（盒须图），展示：时间(time)和账单总金额(total_bill)的分布情况
sns.boxplot(data=df, x='time', y='total_bill')
plt.show()
```

# 小提琴图，适用于箱线图 + 密度图

```python
sns.violinplot(data=df, x='time', y='total_bill', hue='sex', split=True)
plt.show()
```
