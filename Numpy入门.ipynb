```python
import numpy as np
```

# 1.Numpy的属性介绍

## 1.1 创建ndarray对象

```python
# numpy的数组是多维的，且是固定类型的，ndarray，主要用于矩阵运算

# 1.创建数组
arr1 = np.arange(15).reshape(3, 5)
print(arr1)
```

## 1.2 演示numpy的常用属性

```python
print(f'numpy的轴：{arr1.ndim}')
print(f'numpy的维度：{arr1.shape}')
print(f'numpy的元素类型：{arr1.dtype}')
print(f'numpy的元素个数：{arr1.size}')
print(f'numpy的元素字节大小：{arr1.itemsize}')
print(f'numpy的元素类型：{type(arr1)}')
```

# 2. 创建Numpy的ndarray对象

## 2.1 arryar()函数，把python列表转化为ndarray

```python
# 1.创建python列表对象
my_list = [1, 2, 3, 4, 5]
# 2.列表 -> ndarray对象
arr2 = np.array(my_list)
print(arr2)
print(type(arr2))
```

## 2.2 arange()函数，创建ndarrary对象

```python
# arange(起始，结束，步长，dtype=类型)
# 1.创建ndarrary对象
arr3 = np.arange(0, 10, 2)
print(arr3)
print(type(arr3))
print(f'元素类型：{arr3.dtype}')
```

## 2.3 随机数生成ndarrary对象

```python
# 1. random.rand()生成随机小数矩阵 0.0~1.0 左闭右开
arr4 = np.random.rand(3, 5)  # 三行五列
print(arr4)
```

```python
# 2. random.randint(起始值，结束值，size=(行，列)) 生成随机整数矩阵 左闭右开
arr5 = np.random.randint(3, 5, size=(2, 6))  #2行6列
print(arr5)
```

```python
# 3. random.uniform
arr6 = np.random.uniform(3, 9, size=(3, 5))
print(arr6)
```

## 2.4 numpy的类型转换 astype()

```python
arr7 = np.arange(0, 10, 2, dtype=np.int32)
print(arr7)
print(arr7.dtype)
arr8 = arr7.astype(np.float32)
print(arr8)
print(arr8.dtype)
```

## 2.5 创建等比数列

```python
# 起始值（指数），结束值，元素个数，base=（底数）
arr9 = np.logspace(0, 3, 4)
print(arr9)
arr9 = np.logspace(0, 3, 4, base=2, dtype=np.int64)
print(arr9)
```

## 2.6 创建等差数列

```python
arr10 = np.linspace(1, 10, 4)
print(arr10)
```

# 3. numpy常用函数

## 3.1常用函数

```python
# 1. 创建标准正态分布的序列
arr1 = np.random.randn(3, 5)
print(arr1)
```

```python
# 2. 测试基本函数
# print(np.ceil(arr1))  # 向上取整
# print(np.floor(arr1)) # 向下取整
# print(np.rint(arr1))  # 四舍五入
# print(np.abs(arr1))
# print(np.multiply(arr1, arr1))  #矩阵乘法（行列数一致）
# print(arr1 * arr1)    # 同上
# print(np.divide(arr1, arr1))    #矩阵除法
print(np.where(arr1 > 0, 1, -1))  #三目运算符
```

## 3.2 统计函数

```python
arr1 = np.arange(0, 12).reshape(3, 4)
print(arr1)
```

```python
# 累加和
print(np.cumsum(arr1))
# 求和
print(np.sum(arr1, axis=0))  # 0->列
print(np.sum(arr1, axis=1))  # 1->行
```

## 3.3 去重函数

```python
arr1 = np.array([[1, 2, 1], [2, 3, 5]])
print(arr1)
# 去重 返回新副本
arr2 = np.unique(arr1)
print(f'arr1:{arr1}')
print(f'arr2:{arr2}')
```

## 3.4 排序函数

```python
# np.sort() 返回新副本
arr1 = np.array([11, 33, 22, 66, 55])
print(arr1)
arr2 = np.sort(arr1)
print(arr2)

# 数组对象.sort() 修改新副本
print(arr1)
arr1.sort()
print(arr1)
```

# 4. 矩阵运算

## 4.1 减法

```python
arr1 = np.array([10, 20, 30, 40])
arr2 = np.arange(4)
print(arr1)
print(arr2)
print(arr1 - arr2)
# 加法运算
print(arr1 + arr2)
```

## 4.2 矩阵乘法

```python
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
print(arr1)
print(arr2)
# 行列数一致(其实不是矩阵乘法，而是对应位置相乘)
print(arr1 * arr2)
print(np.multiply(arr1, arr2))  # 同上
```

```python
# 矩阵乘法
arr3 = np.array([[1, 2, 3], [4, 5, 6]])
arr4 = np.array([[9, 8], [6, 5], [3, 2]])
print(arr3)
print(arr4)
print(arr3 @ arr4)
print(np.dot(arr3, arr4)) # 效果同上
print(arr3.dot(arr4))  # 效果同上
```
